```python
def describe_pet(animal_type, pet_name):
    """顯示寵物資訊"""
    print(f"\nI have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")
```

```python
#引數名稱呼叫
def describe_pet(animal_type, pet_name):
    print(f"我養了一隻{animal_type}，牠的名字叫做{pet_name.title()}。")

describe_pet(animal_type='貓', pet_name='小菊')
describe_pet(pet_name='小菊', animal_type='貓')
```

## 重要規則 : 混用時 位置引數 一定要前面
# 正確 : 先傳位置引數, 再傳引數名稱

def order_drink(drink, ice='正常冰', sugar='正常糖', size='大杯'):
    """訂購飲料"""
    print(f"我想要一杯{size}的{drink}，{ice}、{sugar}。")


order_drink('珍珠奶茶')  # 錯誤處：原本縮排在函式內，應移到外層，避免遞迴呼叫
order_drink('珍珠奶茶', size='大杯')  # 錯誤處：原本第二個位置引數會指定給 ice，不是 size
order_drink(drink='珍珠奶茶', sugar='半糖', size='大杯')  # 錯誤處：原註解說沒有 sugar 參數，但函式定義中有 sugar
